```python

df_manobs = pd.read_csv('df_all_events_tw_29juillet2026.csv')
df_manobs_soundings = df_manobs[df_manobs.Sounding==True]

df_manobs_iop5_sorel = df_manobs.loc[((df_manobs.IOP=='IOP5') & (df_manobs.site=='Sorel'))]

df_manobs_iop5_sorel = df_manobs_iop5_sorel.copy()
df_manobs_iop5_sorel['rounded_time'] = pd.to_datetime(df_manobs_iop5_sorel['rounded_time'])
df_manobs_iop5_sorel = df_manobs_iop5_sorel.set_index('rounded_time')

# Resample to 1-second frequency and interpolate pr
df_manobs_iop5_sorel_1s = df_manobs_iop5_sorel[['pr']].resample('1s').interpolate(method='time')

def pr_factor(pr):
    pr_in_sim1 = [0.1, 0.5, 1.0, 2.0, 5.0, 10.0]
    pr_out_sim1 =[np.float64(0.09720999999999999), np.float64(0.63879), np.float64(1.47509), np.float64(3.28753), np.float64(8.82431), np.float64(18.29916)]
    if pr>0.1:
        # Interpolate pr
        pr_in = np.interp(pr, pr_out_sim1, pr_in_sim1 )
        return pr_in
    else:
        return pr
    
df_manobs_iop5_sorel_1s_mod_pr = df_manobs_iop5_sorel_1s.copy()
df_manobs_iop5_sorel_1s_mod_pr['pr'] = df_manobs_iop5_sorel_1s_mod_pr['pr'].apply(pr_factor)

shift_s = - 2000 # Snow initialize at 4000 m, and falling at 1 m/s, so it takes 4000 s to reach the ground. So we shift the qitop/nitop by -4000 s

df_manobs_iop5_sorel_1s_mod_pr= (
    df_manobs_iop5_sorel_1s_mod_pr.shift(shift_s)
)
df_manobs_iop5_sorel_1s_mod_pr = (
    df_manobs_iop5_sorel_1s_mod_pr.fillna(0.0)
)

df_manobs_iop5_sorel_1s_mod_pr['qitop'] = df_manobs_iop5_sorel_1s_mod_pr['pr'].apply(lambda x: kin1d_functions.pr_to_N_Q_2(x,all_drop_same_size=False)[0])
df_manobs_iop5_sorel_1s_mod_pr['nitop'] = df_manobs_iop5_sorel_1s_mod_pr['pr'].apply(lambda x: kin1d_functions.pr_to_N_Q_2(x,all_drop_same_size=False)[1])

df_manobs_iop5_sorel_1s_to_write = df_manobs_iop5_sorel_1s_mod_pr[['qitop', 'nitop']]


df_manobs_iop5_sorel_1s_to_write.to_csv('p31d_sip-ffd-mod-notempdep-evolving_T_PR_SOREL_IOP5/df_manobs_iop5_sorel_1s.csv',index=False)

fig,ax=plt.subplots(1,figsize=(6,3))
ax.plot(df_manobs_iop5_sorel_1s.index, df_manobs_iop5_sorel_1s.pr)
ax.plot(df_manobs_iop5_sorel_1s_mod_pr.index, df_manobs_iop5_sorel_1s_mod_pr.pr,color='C1')

print(len(df_manobs_iop5_sorel_1s))
```

```python
def plot_iop5_sorel_obs_pr(df_manobs_iop5_sorel,ax):
    color_list = {'SN and FZRA':'blue','SN and RA':'darkblue','None':'grey','No precip.':'grey','PL':'#1f77b4','FZRA':"#ff0e0e",'FZRA and PL':'#2ca02c','RA':'k','SN and PL':'cyan','SN':'cyan','RA and PL':'darkgreen'}

    marker_list = {'SN and RA':'s','None':'_','No precip.':'_',    'PL':'o'  ,'FZRA':'^','FZRA and PL':'s','SN and FZRA':'s','RA and PL':'s','RA':'s','SN and PL':'s','SN':'s','RA and PL':'s'}

    for v in ['PL','FZRA','FZRA and PL']:#,'RA','SN and PL','SN','RA and PL']:#df_all_events_pr.manobs_with_mix.unique()[1:]:
        mask = df_manobs_iop5_sorel.manobs_with_mix.values == v
        
        
        if (v=='FZRA and PL'):
            markerfacecolor    = '#2ca02c'
            markerfacecoloralt = '#2ca02c'
            markeredgecolor    = '#2ca02c'    
        elif ('and' in v):
            markerfacecolor    = color_list[v.split(' ')[0]]
            markerfacecoloralt = color_list[v.split(' ')[-1]]
            markeredgecolor = 'none'
        else:
            markerfacecolor    = color_list[v]
            markerfacecoloralt = color_list[v]
            markeredgecolor    = color_list[v]
            

        # IOP5 Sorel
        ax.plot(df_manobs_iop5_sorel[mask].index, df_manobs_iop5_sorel[mask].pr, markerfacecolor='white',marker=marker_list[v],linestyle='None',markeredgecolor=markeredgecolor,label=v+' obs.')
        
```

```python
cmap_rhi = mcolors.LinearSegmentedColormap.from_list(
    "highlight_high",
    [
        (0.0, "lightblue"),
        (0.6, "yellow"),
        (0.7, "orange"),
        (1.0, "red")
    ])


bounds =[50,70,90,95,100,101,102,103,105]
# norm_rhi = mcolors.BoundaryNorm(bounds,ncolors=256)

norm_rhi = mcolors.FuncNorm(
    (
        lambda x: np.interp(
            x,
            bounds,
            np.linspace(0,1,len(bounds)),#[0, 0.2, 0.45, 0.65, 0.85, 0.9, 0.94, 0.97, 1]
        ),
        lambda y: np.interp(
            y,
            np.linspace(0,1,len(bounds)),#[0, 0.2, 0.45, 0.65, 0.85, 0.9, 0.94, 0.97, 1],
            bounds
    )
)
)

# norm_rhi = mcolors.TwoSlopeNorm(
#     vmin=50,
#     vcenter=100,
#     vmax=110
# )


cmap1 = plt.cm.Reds(np.linspace(0.5, 1, 200))   # 50-100
cmap2 = plt.cm.Blues(np.linspace(0.5, 1, 200))        # >100

# Combine them
colors = np.vstack((cmap1, cmap2))

cmap_rhi = mcolors.LinearSegmentedColormap.from_list(
    "combined_cmap",
    colors
)

```

```python
# Compute RH, RHI
ds_all["esw"] = xr.apply_ufunc(lambda T:  microphy_p3_qv_sat.microphy_qv_sat.polysvp1(T,  0), ds_all["T"]+273.15 , vectorize=True)
ds_all["esi"] = xr.apply_ufunc(lambda T:  microphy_p3_qv_sat.microphy_qv_sat.polysvp1(T,  1), ds_all["T"]+273.15 , vectorize=True)
ds_all["e_w"] = xr.apply_ufunc(lambda Td: microphy_p3_qv_sat.microphy_qv_sat.polysvp1(Td, 0), ds_all["Td"]+273.15, vectorize=True)

ds_all["rh"]  = ds_all["e_w"]/ds_all["esw"]
ds_all["rhi"] = ds_all["e_w"]/ds_all["esi"]
```
